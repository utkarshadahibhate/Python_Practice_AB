```python
## Note
## 1. functions are object - first class function
## 2. Nested functions
## 3. Function as parameter - when a function is taking another function
##      as a parameter, then the function is called as higher order function.
##      The one who is taking parameter is called as higher order function.
## 4. Returning a function
```

```python
## Returning a Function - Outer function returns inner function
## if a function is returning its inner function, then also that outer function is called as
#   higher order function.

## What is higher order function ?
#   A function which takes a function as parameter or returns a function.
```

```python
def Outer():
    def Inner():
        print('welcome')

    return Inner    ## outer function returning inner function

f = Outer() ## will return inner function to f, so f is a function which is referring to that inner function
f()
```
