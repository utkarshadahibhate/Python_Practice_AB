```python
## Functions as parameters - function takes another function as parameter

def greet():
    print('welcome')

def fun(f): ## function fun taking f as a parameter
    f()     ## f() is also a function called inside fun, function fun thinks that f is a 
            ## function and it is calling it.

fun(greet) ## greet function is passed to function fun, greet goes into f() and its called.
```

```python
def add(x,y):
    return x+y

def sub(x,y):
    return x-y

def arithmetic(f, x, y): ## passing functionality to a function
    return f(x,y)

sum = arithmetic(add,10,5)
diff = arithmetic(sub,10,5)

print('Sum: ',sum,',Diff: ',diff)
```
