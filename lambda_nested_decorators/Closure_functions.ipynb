```python
## Note:
# 1. Functions are object
# 2. Nested functions - (function inside function)
# 3. Function as parameter - (function takes function as parameter)
# 4. Returning a function - (Outer function returns inner function)
# 5. Closure function - (Nested function, Returning function, Inner function access Outer variables)
```

```python
## Closure function means a function having a inner function which is having its variable means
## outer function's variable and outer function will return that inner function.
```

```python
def Outer():            ## outer function
    msg = "welcome"
    def Inner():        ## inner function
        print('+' * 10)
        print(msg)      ## inner function accessing varible of outer function
        print('+' * 10)

    return Inner        ## outer function returning inner function

f = Outer()
f()
```

```python
def Outer(msg):         ## outer function with msg
    def Inner():        ## inner function
        print('+' * 10)
        print(msg)      ## inner function accessing varible of outer function
        print('+' * 10)

    return Inner        ## outer function returning inner function

f = Outer('Python')
f()
```

```python
count = 0       ## global variable
def counter():
    global count
    count += 1      ## accessing and modifying global variable
    return count

print(counter())
print(counter())
```

```python
def get_counter():
    count = 0
    def counter():
        nonlocal count  # if the inner function is modifying the variables of outer function, 
        count += 1      # then it should declare them nonlocal then only it can modify it
        return count
    return counter

c1 = get_counter()
c2 = get_counter()

print(c1(), c1(), c1())
print(c2(), c2(), c2())
```
