```python
# Nested function - function inside another function
def Outer():
    def Inner():    ##nested inner function
        print('Inner')

    print('Outer')
    Inner() ## can only be called through outer function

Outer()
```

```python
Inner() ## will through an error
## so inner function can be called only from inside outer function only
```

```python
## If you have a lot of code inside one of the function, then you can break the task of that
## function into smaller task and write them in the form of inner function.
## to reduce the complexity of a bigger function, you can also write nested function.
```

```python
def totalarea(l, b, h):
    def area(d1, d2):
        return d1 * d2

    return 2 * (area(l,b) + area(b,h) + area(l,h))

print(totalarea(10,5,3))
```
