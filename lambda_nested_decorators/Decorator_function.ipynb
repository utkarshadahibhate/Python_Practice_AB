```python
## Note:
# 1. Function are objects
# 2. Nested functions - function inside function
# 3. Function as parameter - function takes function as parameter
# 4. Returning a function - outer function returns inner function
# 5. Closure function - Nested function, Returning function, Inner access outer variables
# 6. Decorator - Closure function + function as parameter
```

```python
## Decorator - Closure function + function as parameter
```

```python
def Outer(f):
    def Inner():
        print('+' * 10)
        f()
        print('+' * 10)

    return Inner

def display():
    print('Welcome')

r = Outer(display)  ## passing function as a parameter
r()
```

```python
def Outer(f):
    def Inner():
        print('+' * 10)
        f()
        print('+' * 10)

    return Inner

def display():
    print('Welcome')

display = Outer(display)  ## modifying display function, now display is being replaced with that inner
display()                 ## function along with its own functionality, so display function is decorated
```

```python
## A function which can be used as a decorator notation, such function is called as decorator function.
def Outer(f):
    def Inner():
        print('+' * 10)
        f()
        print('+' * 10)

    return Inner

## decorator notation
@Outer          ## by using a notation we can use the decorator function 
def display():
    print('Welcome')

display()
```

```python
## A function which can be used as a decorator notation, such function is called as decorator function.
def decorator(f):
    def Inner():
        print('+' * 10)
        f()
        print('+' * 10)

    return Inner

## decorator notation
@decorator          ## by using a notation we can use the decorator function 
def display():
    print('Welcome')

display()

## so now this display function will be passed to decorator function automatically as a parameter
## and the inner function is returned and then the display function will modify itself as a inner function.
```
