```python
## Caller class - caller class are similar to closure function means instead of defining a closure
## function we  can also  write it as a class
```

```python
## if a class is having call method then it is called a caller class
class Day:
    def __init__(self):
        self.days = {0:'Sunday', 1:'Monday', 2:'Tuesday', 3:'Wednesday', 4:'Thursday', 5:'Friday',
                 6:'Saturday',7:'Sunday'}

    def __call__(self, dayno): ## call method
        return self.days[dayno]

d = Day()
print(d(3)) ## call method can be called directly using the object
## we can treat the object of this clas as a function, object of the class can be used as function.
```
