#### Random access of files

- text files means ASCII data
- binary files means bytes
- but ASCII are also bytes, the reason we are able to access it either sequentially as well as randomly but for accessing randomly we must write 'rb'.
- tell() - will tell the current cursor position
- seek(offset, whence) - how much and where you want to move the cursor

```python
with open('data.txt', mode = 'rb') as file:
    print(file.read(4))
    print(file.tell())     ## will tell the current cursor position

    file.seek(3,0)  ## from the beginning of the file move 3 bytes i.e. ABC and the pointer will move to at D
    print(file.read(2))
```

```python
with open('data.txt', mode = 'rb') as file:
    print(file.read(4))
    print(file.tell())

    file.seek(2,1)      ## from the current cursor position move 2 bytes i.e. GH and the pointer will move to at I
    print(file.read(2))
```

```python
with open('data.txt', mode = 'rb') as file:
    print(file.read(4))
    print(file.tell())

    file.seek(-3,1)     ## to move backward from the current position
    print(file.read(2))
```

```python
with open('data.txt', mode = 'rb') as file:
    print(file.read(4))
    print(file.tell())

    file.seek(-4,2)     ## to move cursor to end of the file
    print(file.read(2))
```
