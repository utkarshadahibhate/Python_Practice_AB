#### Binary vs Text file

#### File types
- Text files - Sequential access
- Binary files - Sequential + Random access

- text files are human readable but binary files are not human readable.
- text files are only in the form of strings but binary files are only in the form of bytes.

```python
i = 10

s = 'Hello'

with open('Sample.txt', mode = 'w') as file:
    file.write(i)           ## will throw an error
    file.write(s)

## In text file, you can write only characters that is in the form of a string in the form of ASCII.
```

```python
i = 10

s = 'Hello'

with open('Sample.txt', mode = 'w') as file:
    file.write(str(i))           
    file.write(s)
```

##### To write a binary file

```python
i = 10

s = 'Hello'

with open('Sample.txt', mode = 'wb') as file:
    file.write(str(i))      ## will throw an error while writing to binary file
    file.write(s)

## it can't store ASCII codes, it need bytes
## it can't store string, it can't store given integer
## binary files must be stored only in the form of bytes
```
