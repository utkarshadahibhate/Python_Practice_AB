```python
## to handle exception python have certain keywords 
# 1. try
# 2. except
# 3. else
# 4. finally
```

```python
a,b = 10,0

c = a//b
print(c)
print('end')
```

```python
## general form of try and except
a , b = 10, 0

try:
    c = a // b    ## ZeroDivisionError
    print(c)
except:
    print('error')

print('end')

## we write down those statements which may cause exception inside the try block
## if there is an error then it will go to except block
## except block will execute if there is only an exception
```

```python
a , b = 10 , 5

try:
    c = a // b
    print(c)
except:
    print('b should not be 0')

print('End of program')
```
