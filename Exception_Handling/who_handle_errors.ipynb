```python
## what about run time error ?
# The developer has developed the software and now user is using that software
## what errors may could occur at user end - 
#   1. User input error
#   2. File access error
#   3. Database connections error
#   4. Internet connection error
#   5. Peripherals error - (connecting printer, camera etc.)

## these are the few reasons for which user may get errors when using the software, so these are runtime errors.

## if the user is doing wrong, we should prompt him that it is invalid.
## Program should prompt him and how the program can prompt him - Developer has written the code already for that.

## Developer knows that this is a suspected place where the user may go wrong, so we will prompt him.
## So in case of runtime errors we can prompt the user and guide him properly.

## At the runtime error the program crashes, but program should not crash but ask the user to rectify that error.

## So the developer developed the software which works perfectly except in the above cases.
#  Except in these cases so these are the exceptional cases.
# So developer should develop the program which handles those exception cases also means random errors.
# that is called as exception handling.

## These are external things that are available to the program, and the program is using it. and they are available at user end.
## So actually all these are called as resources, user will face runtime errors because of the resources.
## He may not be using the resources properly or providing the resources properly to the program.
## that's why it causes runtime errors and we handle those situations.
## that is called as exception handling.

## The programmer has to remove syntax error logical error.
## Make sure that the program is perfect. And also he should write down some extra code for handling these situations.
## And that extra piece of code is called as exception handling.
## He should write down the extra piece of code such that the program works smoothly.
## It should not crash and give proper information to the user so that he can rectify those errors.
```
