```python
## Syntax error - typo error - who will help - Python interpreter
a = input('Enter value of a')
b = input('enter value of b)
## python interpreter helps in correcting the error
```

```python
## Logical error - not getting expected result -we remove these types of error by using debugging

a = input('Enter value of a')
b = input('enter value of b')
if a > b:
    print(a)
else:
    print(a)
```

```python
## Runtime error - an error occured during the execution of the program, that error is called as runtime error
## how to identify runtime error in Python ?
## It will show Traceback means in which line it has given an error where it has stopped suddenly or crashed.
a = input('Enter value of a')
b = input('enter value of b')

c = a//b
print(c)
## Program has executed and suddenly stoppedin line number 5 because this division cannot be done
## so this is nothing but runtime error
```
