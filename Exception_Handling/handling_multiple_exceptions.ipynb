```python
## Handling multiple exceptions
```

```python
L = [10,20,30,40,50]

index = 9
print(L[index])

print('End of program')
```

```python
L = [10,20,30,40,50]

index = 'abc'
print(L[index])

print('End of program')
```

```python
## There are two possibilites of exception - wrong index or not giving integer value
L = [10,20,30,40,50]

try:
    index = 9
    print(L[index])
except IndexError:
    print('Invalid Index')
except TypeError:
    print('Index should be int')
except:
    print('Some error')
print('End of program')

## for one try block we have multiple except blocks and each except block is handling a specific exception
```

```python
L = [10,20,30,40,50]

try:
    index = 'abc'
    print(L[index])
except IndexError:
    print('Invalid Index')
except TypeError:
    print('Index should be int')
except:
    print('Some error')
print('End of program')
```

```python
L = [10,20,30,40,50]

try:
    index = int('abc')
    print(L[index])
except IndexError:
    print('Invalid Index')
except TypeError:
    print('Index should be int')
except:
    print('Some error')
print('End of program')
```

```python
## Exception handling for index error
L = [10,20,30,40,50]

try:
    index = 9
    print(L[index])
    #print('End of try')
except IndexError as msg:   ## msg refers a exception object
    print(msg)

print('End of program')
```

```python
## Exception handling for index error
L = [10,20,30,40,50]

try:
    index = 'abc'
    print(L[index])
    #print('End of try')
except (IndexError,TypeError) as msg:   ## msg refers a exception object
    print(msg)

print('End of program')
```

```python
## Exception handling for index error
L = [10,20,30,40,50]

try:
    index = int('abc')
    print(L[index])
    #print('End of try')
except (IndexError,TypeError) as msg:   ## msg refers a exception object
    print(msg)

print('End of program')
```
