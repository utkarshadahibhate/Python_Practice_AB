#### Modes of Opening

- r - read from a file
- w - write to a file
- a - append a file
- r+ - read & write
- w+ - write & read
- a+ - append & read
- x - write in a new file only

#### related to binary files
- rb - read binary file
- wb - write binary file
- ab - append a binary file

```python
file = open('My_data', mode = 'w')    ## if the file exists then it will write in that,
                                        ## if the file does not exists then it will create a file and write in it
st = 'abcdefghijklmn'
file.write(st)
file.close()
```

```python
## will overwrite the existing file
file = open('My_data', mode = 'w')
st = '123456789'
file.write(st)
file.close()
```

```python
## will append into the existing file
file = open('My_data', mode = 'a')
st = '\nABCDEFGHIJKLMN'
file.write(st)
file.close()
```

```python
## using read and append mode 'r+'
file = open('My_data', mode = 'r+')
print(file.read())  ## first it will read the file and move the cursor to the end
st = '\npqrstuvwxyz'
file.write(st)      ## will append the string at the end, from where the cursor started
file.close()
```

```python
## using the write and read mode 'w+'
file = open('My_data', mode = 'w+')
st = '\nABCDEFGHIJKLMNOPQRST'
file.write(st)
file.seek(0,0)      ## to move the cursor at the beginning
print(file.read())
file.close()
```
