#### Operations on file

#### functions only for reading
- read(size)
- readline()
- readlines()

```python
file = open('data.txt', mode = 'r')
data = file.read()
print(data)
file.close()
```

```python
file = open('data.txt', mode = 'r')
data = file.read(10)    ## will read just 10 letters/bites of the file
print(data)
file.close()
```

```python
file = open('data.txt', mode = 'r')
data = file.read(10)    ## will read just 10 letters/bites of the file
print(data)

data = file.read(10)    ## will read just 10 letters/bites of the file
print(data)

file.close()
```

```python
file = open('data.txt', mode = 'r')
data = file.readline()    ## will read just one line in the file
print(data)
file.close()
```

```python
file = open('data.txt', mode = 'r')
data = file.readlines()    ## will read the file line by line by giving list of lines
print(data)
file.close()
```

#### functions only for writing
- write()
- writelines()
- truncate()

```python
## using writelines() to write a list of words into a file
file = open('My_data', 'w')
lst = ['one\n','two\n','three\n','four\n','five\n']
file.writelines(lst)
file.close()
```

#### cursor position
- tell() - tells the position of the cursor
- seek(offset, 0/1/2) - (offset = how much you want to move),(0 = beginning of the file, 1 = current postion of the cursor in the file, 2 = at the end of the file)

- everytime when we open a file, then it should be closed at the end otherwise it will cause an error, but if somehow you forget to close it then also it will cause an error
- for this reason there is an another method of opening a file in which the file gets automatically closed, so there is no worries for closing the file after opening it
- open with('file_name', mode = '') as file:
-       indented code block

```python
with open('data.txt', mode = 'r') as file:
    data = file.read()
    print(data)
```
