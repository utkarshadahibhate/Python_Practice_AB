```python
## generator function to yield next month
def next_month():
    count = 1
    while True:
        name = cal.month_name[count]
        yield name
        count = count % 12 + 1
```

```python
month = next_month()
print(next(month))
print(next(month))
```

```python
for mon in next_month(): ##calling generator function using for loop (its an infinite function)
    print(mon)
```
