```python
## if try block is successfully executed, then else block is executed
## if there is not exception then else block will also execute
## if there are any dependent statement inside the try block which may not cause exception
## you should put them in else block so they will execute if there is no exception

a = 10
b = 5

try:
    c = a // b
    ##print(c)        ## dependent statement
except:
    print('b should not b 0')
else:
    print(c)        ## dependent statement (dependent on try block)
```
