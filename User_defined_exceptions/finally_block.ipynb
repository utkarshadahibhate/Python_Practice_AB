```python
## finally block
## useful only inside the function
## finally block is executed even if the function is returning or raising an exception

def fun():

    try:
        x = int('256')
        return x

    except Exception as e:
        raise e

    print('End of function')    ## will not execute

## handling the exception in the form of a variable
## now whatever the exception is raised, it will become an object e and will raise that e.
res = fun()
print(res)
```

```python
def fun():

    try:
        x = int('256')
        return x

    except Exception as e:
        raise e
    
    finally:
        print('End of function')    ## now it will execute
    
res = fun()
print(res)
## finally block is definitely executed even if the function is returning or raising an exception
```
