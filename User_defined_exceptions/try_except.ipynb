```python
a = 10
b = 0

try:
    c = a//b
    print(c)
except:
    print('b should not be 0')

print('End of program')
```

```python
## We can use if and else and we can do the same thing.
a = 10
b = 0

if b != 0:
    c = a//b
    print(c)
else:
    print('b should not be 0')

print('End of program')
```

```python
## Exception handling is useful in between the function
# if you are writing simple main program then you don't require exception handling
# mostly unless you are handling built in exception raised by Python
# But when we are writing a program in the form of functions, then we need exception handling.
```

```python
def div(a, b):

    if b != 0:
        c = a // b
        return c
    else:
        return -1

res = div(10,5)
print(res)
```

```python
def div(a, b):

    if b != 0:
        c = a // b
        return c
    else:
        return -1

res = div(10,0)
print(res)
```

```python
def div(a, b):

    if b != 0:
        c = a // b
        return c
    else:
        return -1

res = div(10,0)
if res != -1:
    print(res)
else:
    print("Division by Zero")
```

```python
def div(a, b):

    if b != 0:
        c = a // b
        return c
    else:
        return -1

res = div(10,-10)   ## will result in -1, interpreting that there is an error but actually that is a result
if res != -1:
    print(res)
else:
    print("Division by Zero")
## therefore we can't make -1 as a code
# then what ? some other number, again that can be a result
## So the problem in giving error codes is that whatever the code you select possibly that is an answer
```

```python
## If a function is able to obtain the result, it should return
## If it cannot get the result, it should raise an exception
def div(a, b):
    if b != 0:
        c = a // b
        return c
    else:
        raise ZeroDivisionError

try:
    res = div(10,0)
    print(res)
except:
    print('Division by Zero')

## When the function is being called from here it should be ready for two things
## Whenever you are calling a function, you should be ready for two things.
## One is whatever the result is returned by a function, you should take it in a variable or 
## else funtions may raise exception, so you should be ready to handle that exception also.
## How to handle the exception - try and except are required
```
