### Nested try and except

```python
L = [10,20,30,40,50]
index = int(input('Enter Index'))
print(L(index))
```

```python
L = [10,20,30,40,50]

try:
    index = int(input('Enter index'))   ## Causes ValueError
    print(L[index])             ## Causes IndexError
except ValueError as e:
    print(e)
except IndexError as e:
    print(e)
```

```python
L = [10,20,30,40,50]

try:
    try:
        index = int(input('Enter Index'))
    except ValueError as e:
        print(e)

    print(L[index])
except IndexError as e:
    print(e)
```

```python
L = [10,20,30,40,50]

try:
    try:
        index = int(input('Enter Index'))
    except ValueError as e:
        print(e)

    print(L[index])
except IndexError as e:
    print(e)
```

```python
L = [10,20,30,40,50]

try:
    try:
        index = int(input('Enter index'))
    except ValueError as e:
        print(e)

    print(L[index])

except IndexError as e:
    print(e)

except Exception as e:
    print(e)
```
