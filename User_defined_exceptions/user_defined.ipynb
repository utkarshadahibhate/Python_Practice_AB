## User Defined exception

```python
## When you write your own exception it should also inherit from exception class directly or indirectly.
##
```

```python
class NegativeError(Exception): ## inheriting from exception class
    pass
```

```python
def area(length,breadth):
    if length >= 0 and breadth >= 0:
        a = length * breadth
        return a
    else:
        raise NegativeError('-ve dimension')

res = area(20,-10)
print(res)
```

```python
class NegativeError(Exception):
    def __init__(self):     ## constructor of the class
        self.msg = "-ve Dimensions"     ## defining error msg
    def __str__(self):      ## string function, when you want to print the object of this class, whatever is 
                            ## returned from here that will be printed
        return self.msg

def area(length, breadth):
    if length >= 0 and breadth >= 0:
        a = length * breadth
        return a
    else:
        raise NegativeError()


print(area(-5, 3))  ## will give the same error with our defined exception
```
