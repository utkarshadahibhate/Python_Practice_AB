```python
## functions for different cases
filename = 'Journal.txt'

def write_entry():
    entry = input('Write your journal entry: ') + '\n'
    with open(filename, mode = 'w') as file:
        file.write(entry)

def read_all():
    with open(filename, mode = 'r') as file:
        print('\nAll Journal Entries: ')
        print(file.read())

def search():
    keyword = input("Enter keyword to search: ")

    with open(filename, mode = 'r') as file:
        all_lines = file.readlines()
        for line in all_lines:
            if keyword.lower() in line.lower():
                print(line.strip())
                return                      ## just simply return to the function after printing
    print('No matching entries')        ## after the loop ends without any matching keyword

## main menu function with cases

def main_menu():

    while True:
        print('\n====== File Journal Menu ========')
        print('1. Add entry')
        print('2. Read all entries')
        print('3. Search entries')
        print('4. Exit')

        choice = input('Choose option (1-4): ')

        if choice == '1':
            write_entry()
        elif choice == '2':
            read_all()
        elif choice == '3':
            search()
        elif choice == '4':     ## exit condition
            print('Goodbye!')
            break                   
        else:
            print('Invalid choice.')

if __name__ == "__main__":
    main_menu()
```
