```python
## Built-in functions
# abs(x,/) - return the absolute value (positive magnitude) of number.
# pow(base, exp, mode = None,/)
# round(number, ndigits = None)
# divmode(a, b,/)
# min(iterable,*,key=None, default=None)
# max(iterable,*,key=None, default=None)
# sum(iterable, start = 0,/)
# eval(expression, globals=None, locals=None)
```

```python
print(abs(-70.12))
print(abs(-70))
print(abs(3 + 4j))
```

```python
## pow examples
print(pow(2, 3))
print(pow(10, 2))   ## output : 100 (10^2)
print(pow(10, -2))   ## output : 0.01 (1/10^2)
print(pow(10, 2, 3)) ## output: 1 (100 mod 3)
(print(100**2 % 3))
```

```python
## round examples
print(round(4.4))
print(round(4.6))
print(round(4.5))   # (banker's rounding)
print(round(5.5))   #(banker's rounding)
print(round(3.54321, 2))   # (round to 2 decimal places) 
print(round(9734, -1))     # (round to nearest tens) 
print(round(9734, -2))     # (round to nearest hundreds)
```

```python
## divmod examples
print(divmod(10, 3)) ## output: (3,1) -> quotien=3, remainder=1
print(divmod(61,7))
```

```python
## min with key and default arguments
numbers = [-10, 3, 7, -2, -5]
print(min(numbers))
print(min(numbers, key=abs)) ## minimum by absolute value
print(min([], default="empty")) ## output: 'Empty (no error on empty iterable)
```

```python
## max with key argument
word = ['apple','banana','cherry','blueberry']
print(max(word, key=len))
```

```python
## sum examples
print(sum([1,2,3,4,5]))
print(sum([1,2,3,4,5], start=20))
```

```python
## eval examples
global_dict = {'x':10,'y':20}
local_dict = {'z':5}
expr = "x + y + z"
print(eval(expr, global_dict, local_dict))
```
